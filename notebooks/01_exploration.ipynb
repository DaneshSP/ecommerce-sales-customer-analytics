```python
query = """
SELECT 
    strftime('%Y-%m', o.order_purchase_timestamp) AS month,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
WHERE o.order_status = 'delivered'
GROUP BY month
ORDER BY month;
"""

monthly_revenue = pd.read_sql(query, engine)
monthly_revenue
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(monthly_revenue['month'], monthly_revenue['total_revenue'], marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (BRL)')
plt.tight_layout()
plt.show()
```

```python
query = """
SELECT 
    p.product_category_name,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    COUNT(oi.order_item_id) AS total_items_sold
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
GROUP BY p.product_category_name
ORDER BY total_revenue DESC
LIMIT 10;
"""

top_categories = pd.read_sql(query, engine)
top_categories
```

```python
query = """
SELECT 
    ct.product_category_name_english,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    COUNT(oi.order_item_id) AS total_items_sold
FROM order_items oi
JOIN products p ON oi.product_id = p.product_id
JOIN category_translation ct ON p.product_category_name = ct.product_category_name
GROUP BY ct.product_category_name_english
ORDER BY total_revenue DESC
LIMIT 10;
"""

top_categories_en = pd.read_sql(query, engine)
top_categories_en
```

```python
query = """
SELECT 
    o.customer_id,
    MAX(o.order_purchase_timestamp) AS last_purchase_date,
    COUNT(DISTINCT o.order_id) AS frequency,
    ROUND(SUM(oi.price), 2) AS monetary
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
WHERE o.order_status = 'delivered'
GROUP BY o.customer_id;
"""

rfm_raw = pd.read_sql(query, engine)
rfm_raw.head()
```

```python
rfm_raw['last_purchase_date'] = pd.to_datetime(rfm_raw['last_purchase_date'])

reference_date = rfm_raw['last_purchase_date'].max()

rfm_raw['recency'] = (reference_date - rfm_raw['last_purchase_date']).dt.days

rfm_raw.head()
```

```python
rfm_raw['r_score'] = pd.qcut(rfm_raw['recency'], 4, labels=[4,3,2,1])
rfm_raw['f_score'] = pd.qcut(rfm_raw['frequency'].rank(method='first'), 4, labels=[1,2,3,4])
rfm_raw['m_score'] = pd.qcut(rfm_raw['monetary'], 4, labels=[1,2,3,4])

rfm_raw['rfm_score'] = rfm_raw['r_score'].astype(str) + rfm_raw['f_score'].astype(str) + rfm_raw['m_score'].astype(str)

rfm_raw.head()
```

```python
def segment_customer(row):
    if row['r_score'] == '4' and row['f_score'] == '4':
        return 'Champions'
    elif row['r_score'] in ['3','4'] and row['f_score'] in ['3','4']:
        return 'Loyal Customers'
    elif row['r_score'] in ['3','4']:
        return 'Recent Customers'
    elif row['r_score'] in ['1','2'] and row['f_score'] in ['3','4']:
        return 'At Risk'
    else:
        return 'Lost/Low Value'

rfm_raw['segment'] = rfm_raw.apply(segment_customer, axis=1)

rfm_raw['segment'].value_counts()
```

```python
rfm_raw.to_sql('customer_rfm', engine, if_exists='replace', index=False)
rfm_raw.to_csv('../data/cleaned/customer_rfm.csv', index=False)

print("RFM data saved")
```

```python
query = """
SELECT 
    o.customer_id,
    o.order_id,
    o.order_purchase_timestamp
FROM orders o
WHERE o.order_status = 'delivered';
"""

cohort_data = pd.read_sql(query, engine)
cohort_data['order_purchase_timestamp'] = pd.to_datetime(cohort_data['order_purchase_timestamp'])

cohort_data.head()
```

```python
cohort_data['order_month'] = cohort_data['order_purchase_timestamp'].dt.to_period('M')

cohort_data['cohort_month'] = cohort_data.groupby('customer_id')['order_purchase_timestamp'].transform('min').dt.to_period('M')

cohort_data.head()
```

```python
cohort_data['cohort_index'] = (cohort_data['order_month'] - cohort_data['cohort_month']).apply(lambda x: x.n)

cohort_pivot = cohort_data.groupby(['cohort_month', 'cohort_index'])['customer_id'].nunique().reset_index()

cohort_table = cohort_pivot.pivot(index='cohort_month', columns='cohort_index', values='customer_id')

cohort_table.head()
```

```python
cohort_size = cohort_table.iloc[:,0]

retention_table = cohort_table.divide(cohort_size, axis=0).round(3) * 100

retention_table.head()
```

```python
retention_table.to_csv('../data/cleaned/retention_table.csv')
cohort_data.to_csv('../data/cleaned/cohort_data.csv', index=False)

print("Cohort/retention data saved")
```
